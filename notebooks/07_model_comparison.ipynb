Итоговая таблица и графики по всем моделям.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../reports/results/metrics_summary.csv')
df
```

```python
# Main comparison table
display_cols = ['model', 'mIoU', 'Dice_tooth', 'Dice_bone', 'Dice_cortical',
                'AUROC_macro', 'F1_macro', 'FPS']
cols_present = [c for c in display_cols if c in df.columns]
print(df[cols_present].to_string(index=False))
```

```python
# Dice per class bar chart
dice_cols = ['Dice_tooth', 'Dice_bone', 'Dice_cortical']
dice_cols = [c for c in dice_cols if c in df.columns]
if dice_cols:
    melt = df.melt(id_vars='model', value_vars=dice_cols,
                   var_name='Class', value_name='Dice')
    melt['Class'] = melt['Class'].str.replace('Dice_', '')
    fig, ax = plt.subplots(figsize=(11, 5))
    sns.barplot(data=melt, x='model', y='Dice', hue='Class', ax=ax)
    ax.set_ylim(0, 1)
    ax.set_title('Dice Score per Class across Architectures')
    plt.tight_layout()
    plt.savefig('../reports/figures/dice_per_class_bars.png', dpi=150)
    plt.show()
```

```python
# Best model selection
if 'mIoU' in df.columns:
    best = df.loc[df['mIoU'].idxmax()]
    print(f'Best model by mIoU: {best["model"]}  (mIoU={best["mIoU"]:.4f})')
    print('Reason for selection:')
    print('  - Highest mIoU = best overall segmentation quality')
    print('  - Dice_cortical is critical: small class, directly impacts classification')
    print('  - FPS must be >1 for practical use in the Gradio app')
```
